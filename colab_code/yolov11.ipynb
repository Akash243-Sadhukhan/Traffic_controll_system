```python
import zipfile
import os

# Define the path to your zip file
zip_file_path = '/content/drive/MyDrive/ml_project_dataset/IDDDetectionsYOLODataset/archive(3).zip' # Replace with the actual path to your zip file

# Define the directory where you want to extract the contents
extraction_dir = '/content/extracted_files' # Replace with your desired extraction directory

# Create the extraction directory if it doesn't exist
os.makedirs(extraction_dir, exist_ok=True)

# Unzip the file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extraction_dir)

print(f"Files extracted to: {extraction_dir}")
```

```python
from ultralytics import YOLO
import os

# Load a COCO-pretrained YOLO11n model
model = YOLO("yolo11n.pt")

# Update Ultralytics settings to point to the correct dataset directory
dataset_path = "/content/extracted_files/Vehicle_Detection"
os.system(f"yolo settings dataset_dir={dataset_path}")


# Train the model on the COCO8 example dataset for 100 epochs
results = model.train(data="/content/extracted_files/Vehicle_Detection/vehicles.yaml", epochs=100, imgsz=640)

# Run inference with the YOLO11n model on the 'bus.jpg' image
results = model("/content/extracted_files/Vehicle_Detection/results")
```
